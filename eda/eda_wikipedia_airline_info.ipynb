# Airline EDA

```python
import pandas as pd

URL = "https://en.wikipedia.org/wiki/List_of_airline_codes"

tables = pd.read_html(URL, storage_options={"User-Agent": "eda-script/0.1"})
airline = pd.concat(
    [t for t in tables if "IATA" in t.columns],
    ignore_index=True,
)

airline.info()
airline.head(100)
```
