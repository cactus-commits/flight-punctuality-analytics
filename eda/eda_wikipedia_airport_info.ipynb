## EDA airport info

```python
import pandas as pd

URL = "https://en.wikipedia.org/wiki/List_of_international_airports_by_country"

tables = pd.read_html(URL, storage_options={"User-Agent": "eda-script/0.1"})
airports = pd.concat(
    [t for t in tables if "IATA Code" in t.columns],
    ignore_index=True,
)

airports.info()
airports.head(100)
```
