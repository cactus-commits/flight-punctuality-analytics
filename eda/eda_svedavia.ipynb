## Fetch API

```python
import os
import requests
import pandas as pd
from dotenv import load_dotenv

load_dotenv()

base_url = os.getenv("BASE_URL")   # t.ex. https://api.swedavia.se/flightinfo/v2
api_key = os.getenv("FLIGHT_KEY")

url = f"{base_url}/ARN/arrivals/2026-10-08"
headers = {
    "Ocp-Apim-Subscription-Key": api_key,
    "Accept": "application/json",
}

response = requests.get(url, headers=headers, timeout=10)
response.raise_for_status()   

data = response.json()
df = pd.json_normalize(data["flights"])
print(df.head())
```
