## Function for airports (svedavia) & Dates as Parameters

```python
# 1. Get BASE_URL & API KEY from dotenv file 
BASE_URL = os.getenv("BASE_URL")
API_KEY = os.getenv("FLIGHT_KEY")
```

```python
# 2. Create an array with the Svedavia Airports 
SWE_AIRPORTS = {
    "ARN": "Stockholm Arlanda Airport",
    "BMA": "Bromma Stockholm Airport",
    "GOT": "Göteborg Landvetter Airport",
    "MMX": "Malmö Airport",
    "LLA": "Luleå Airport",
    "UME": "Umeå Airport",
    "OSD": "Åre Östersund Airport",
    "VBY": "Visby Airport",
    "RNB": "Ronneby Airport",
    "KRN": "Kiruna Airport",
}
```

**Source used:** https://www.geeksforgeeks.org/python/get-yesterdays-date-using-python/ 

```python
# 3. Create a parameter for the dates to fetch 

# Import date and timedelta class
# from datetime module
from datetime import date
from datetime import timedelta

# Array with the days [today -1, today-2, today-3] 
DATES = [date.today(), date.today() - timedelta(days=1), date.today() -
         timedelta(days=2)]

#print the first day in the array (yesterday)
print(DATES[0])


# print all the dates 
for date in DATES:
    print(date)

```

```python
# 5. Create a header - request - response 
# 6. Loop 
# For each AIRPORT 
# For each DATE

def get_flight_data(direction):
    # Empty list to store the flights in 
    all_flights = []

    headers = {
        "Ocp-Apim-Subscription-Key": API_KEY,
        "Accept": "application/json",
    }
    for airport in SWE_AIRPORTS: 
        for date in DATES: 
            url = f"{BASE_URL}/{airport}/{direction}/{date}"
            response = requests.get(url, headers=headers)
            response.raise_for_status()
            for flight in response.json()["flights"]:
                all_flights.append({**flight, "Swedavia_airport_name": airport, "Date_fetched": str(date)}) # adderade datum för det hämtade datumet också för att få in det i DF 
    return all_flights
```

```python
# 7. Print results 

df_arrivals = pd.json_normalize(get_flight_data("arrivals"), sep="_")
```

```python
df_departures = pd.json_normalize(get_flight_data("departures"), sep="_")
```
